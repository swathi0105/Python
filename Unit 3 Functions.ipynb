###TYPES OF ARGUMENTS


1.   Positional Argument
2.   Keyword Argument
3.   Default Argument
4.   (*args) Arbitrary Positional Argument
5.   (**kwargs) Keyword Positional Argument



```python
def add(a,b):
  print("Add",a+b)
#Positional Argument
print(add(5,10))
#Keyword Argument and Default Argument
print(add(a=20,b=10))
```

####4. Arbitrary Positional Arguments (*args)
       

*   accepts any number of extra positional arguments as a tuple


```python
def add(*args):
  c=sum(args)
  print("Add:",c)

add(1,2,3,4,5,10)
```

```python
def add(**kwargs):
  c=sum(kwargs.values())
  print("Add:",c)

add(a=10, b=5, extra=5)
```

### Tower of Hanoi(Recursion)

```python
def hanoi(n, source, destination, helper):
  if n == 1: # Base case
    print(f"Move disk 1 from {source} to {destination}")
    return
  hanoi(n - 1, source, helper, destination) #Recursion step1
  print(f"Move disk {n} from {source} to {destination}")
  hanoi(n - 1, helper, destination, source) #Recursion step2
hanoi(3, 'A', 'C', 'B')
```

### Student Grade Management Program

```python
name = input("Enter your name:")

def total(sub1,sub2,sub3):
  total=sub1+sub2+sub3     #local variable
  return total             #return

def average(total):        #passing total as parameters
  average=total/3
  return average

def grade(average):
  if average >= 90:
    print("Grade A")
  elif average >= 80:
    print("Grade B")
  elif average >= 70:
    print("Grade C")

t = total(100, 100, 100)
print("Total",t)
a= average(t)
print("Average",a)
grade(a)

```
