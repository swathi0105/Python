1. SIMPLE IF

```python
n=int(input("Enter a number "))
if (n>0):
  print("It is a whole number")
```

2. If-Else statement

```python
age=int(input("Enter your age :"))

if (age>=18):
  print("Eligibility to vote")
else:
  print("Not eligible to vote")
```

3. Laddef if statement - It contains morethan 1 condition. It has a special keyword elif to use more conditions.


```python
m = int(input("Enter your mark :"))

if m >= 90:
    print("Grade: A")
elif m >= 75:
    print("Grade: B")
elif m >= 50:
    print("Grade: C")
else:
    print("Grade: F")
```

4. Nested if - It contains an if-else within an if-else

```python
a = int(input("Enter A value :"))
b = int(input("Enter B value :"))
c = int(input("Enter C value :"))

# Step 1: Comparing 'a' against 'b'
if a > b:
    # 'a' beat 'b'! Now 'a' only needs to beat 'c'
    if a > c:
        print("A is the greatest!")
    else:
        print("C is the greatest!")

else:
    # 'b' beat 'a'! Now 'b' only needs to beat 'c'
    if b > c:
        print("B is the greatest!")
    else:
        print("C is the greatest!")
```

Grestest of three using Ladder-if

```python
a = int(input("Enter A value :"))
b = int(input("Enter B value :"))
c = int(input("Enter C value :"))

if (a>b) and (a>c):
  print("A is the greatest number!")
elif (b>a) and (b>c):
  print("B is the greatest number!")
else :
  print("C is the greatest number!")
```

Greatest of three using max() ( Its an aggregate function )

```python
a = int(input("Enter A value :"))
b = int(input("Enter B value :"))
c = int(input("Enter C value :"))

print("Greatest of three",max(a,b,c))
```
