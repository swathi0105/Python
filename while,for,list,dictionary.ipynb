##List Manipulations

image.png image.png

```python
# Base list example
fruits = ["apple", "banana", "cherry"]

print("--- Original List ---")
for item in fruits:
    print(item)
print()

# 1. append(): Adds a single item to the end
fruits.append("dragonfruit")
print("1. append('dragonfruit'):", fruits)

# 2. extend(): Appends elements from another list to the end
fruits.extend(["elderberry", "fig"])
print("2. extend(['elderberry', 'fig']):", fruits)

# 3. insert(): Inserts an item at a specific index
fruits.insert(1, "blueberry")
print("3. insert(1, 'blueberry'):", fruits)

# 4. remove(): Removes the first occurrence of a value
fruits.remove("banana")
print("4. remove('banana'):", fruits)

# 5. pop(): Removes and returns item at index 2
removed_item = fruits.pop(2)
print("5. pop(2): Removed", removed_item, "->", fruits)

# 6. index(): Returns the index of the first matching item
idx = fruits.index("apple")
print("6. index('apple'):", idx)

# 7. count(): Returns total occurrences of a value
cnt = fruits.count("apple")
print("7. count('apple'):", cnt)

# 8. sort(): Sorts items in-place in alphabetical order
fruits.sort()
print("8. sort():", fruits)

# 9. reverse(): Reverses list items in-place
fruits.reverse()
print("9. reverse():", fruits)

# 10. copy(): Creates a shallow copy of the list
fruits_backup = fruits.copy()
print("10. copy():", fruits_backup)

# 11. len(): Returns total element count
print("11. len(fruits):", len(fruits))

# 12. min(): Returns the smallest item (alphabetically)
print("12. min(fruits):", min(fruits))

# 13. max(): Returns the largest item (alphabetically)
print("13. max(fruits):", max(fruits))

# 14. sum(): Calculates total sum of numerical items
prices = [10, 20, 30, 40]
print("14. sum([10, 20, 30, 40]):", sum(prices))

# 15. clear(): Removes all items from the list
fruits.clear()
print("15. clear():", fruits)
```

```python
# 1. fromkeys(): Create a new dictionary from a sequence of keys with a default value
keys_list = ["math", "physics", "chemistry"]
scores = dict.fromkeys(keys_list, 0)
print("1. fromkeys():", scores)

# Initialize a base dictionary for further operations
student = {"name": "Alice", "age": 20, "course": "CS", "grade": "A"}

# 2. keys(): Retrieve a view object of all dictionary keys
print("2. keys():", list(student.keys()))

# 3. values(): Retrieve a view object of all dictionary values
print("3. values():", list(student.values()))

# 4. items(): Retrieve a view object of (key, value) pairs
print("4. items():", list(student.items()))

# 5. get(): Safely fetch value by key without throwing KeyError
print("5. get('name'):", student.get("name"))
print("   get('city', 'Not Found'):", student.get("city", "Not Found"))

# 6. update(): Merge key-value pairs from another dictionary
student.update({"gpa": 3.9, "status": "Active"})
print("6. update():", student)

# 7. setdefault(): Returns value if key exists; inserts key with default if missing
city = student.setdefault("city", "New York")
print("7. setdefault():", student)

# 8. pop(): Remove specific key and return its associated value
removed_age = student.pop("age")
print("8. pop('age'): Removed", removed_age, "->", student)

# 9. popitem(): Remove and return the last inserted (key, value) pair
last_pair = student.popitem()
print("9. popitem(): Removed", last_pair, "->", student)

# 10. copy(): Create a shallow copy of the dictionary
student_backup = student.copy()
print("10. copy():", student_backup)

# 11. len(): Return total number of key-value pairs
print("11. len():", len(student))

# 12. in Operator: Check if a key exists in the dictionary
print("12. 'name' in student:", "name" in student)
print("    'age' in student:", "age" in student)

# 13. del Statement: Delete a specific key-value pair
del student["course"]
print("13. del student['course']:", student)

# 14. sorted(): Return a sorted list of dictionary keys
print("14. sorted() keys:", sorted(student.keys()))

# 15. clear(): Remove all key-value pairs from the dictionary
student.clear()
print("15. clear():", student)
```
