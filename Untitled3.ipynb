### Tower of Hanoi(Recursion)

```python
def hanoi(n, source, destination, helper):
  if n == 1: # Base case
    print(f"Move disk 1 from {source} to {destination}")
    return
  hanoi(n - 1, source, helper, destination) #Recursion step1
  print(f"Move disk {n} from {source} to {destination}")
  hanoi(n - 1, helper, destination, source) #Recursion step2
hanoi(3, 'A', 'C', 'B')
```

### Student Grade Management Program

```python
name = input("Enter your name:")

def total(sub1,sub2,sub3):
  total=sub1+sub2+sub3     #local variable
  return total             #return

def average(total):        #passing total as parameters
  average=total/3
  return average

def grade(average):
  if average >= 90:
    print("Grade A")
  elif average >= 80:
    print("Grade B")
  elif average >= 70:
    print("Grade C")

t = total(100, 100, 100)
print("Total",t)
a= average(t)
print("Average",a)
grade(a)

```
